```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("diamonds.csv")

# Display the first five rows
df.head()
```

```python
# Check dataset size
print("Dataset shape:", df.shape)
```

```python
# Check column names and data types
df.info()
```

```python
# Generate summary statistics
display(df.describe())
```

```python
# Check missing values
print(df.isnull().sum())
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# 1. Check the number of rows and columns
print("Dataset Shape:", df.shape)
```

```python
# 2. Check column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python

# Count missing values in each column
print("Missing Values:")
display(df.isnull().sum())
```

```python

# Check data types
print("\nData Types:")
display(df.dtypes)
```

```python
# Check for zero or negative physical dimensions
dimension_cols = ["x", "y", "z"]
```

```python
print("\nZero or Negative Dimensions:")
display((df[dimension_cols] <= 0).sum())
```

```python

# Keep the original data unchanged
diamonds = df.copy()
```

```python
# Record the original number of rows
original_rows = len(diamonds)
print("Original rows:", original_rows)
print("Working copy created successfully.")
```

```python

# Count duplicates before cleaning
duplicates_before = diamonds.duplicated().sum()

# Remove exact duplicate rows
diamonds = diamonds.drop_duplicates().copy()

# Report the result
duplicates_after = diamonds.duplicated().sum()

print("Duplicates before:", duplicates_before)
print("Rows removed:", original_rows - len(diamonds))
print("Duplicates after:", duplicates_after)
print("Cleaned dataset shape:", diamonds.shape)
```

```python
# Check missing values
print("Missing values per column:")
display(diamonds.isnull().sum())

# Check invalid physical dimensions
dimension_cols = ["x", "y", "z"]

print("Zero or negative dimensions:")
display((diamonds[dimension_cols] <= 0).sum())
```

```python

# Check the cleaned dataset
print("Final shape:", diamonds.shape)
print("Remaining duplicates:", diamonds.duplicated().sum())
print("Total missing values:", diamonds.isnull().sum().sum())

# Preview the cleaned data
display(diamonds.head())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titleweight"] = "bold"
```

```python

plt.figure(figsize=(10, 6))

sns.histplot(data=diamonds, x="price", bins=40, kde=True)

plt.title("Distribution of Diamond Prices")
plt.xlabel("Price")
plt.ylabel("Number of Diamonds")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.histplot(data=diamonds, x="carat", bins=40, kde=True)

plt.title("Distribution of Diamond Carat Weight")
plt.xlabel("Carat Weight")
plt.ylabel("Number of Diamonds")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=diamonds,
    x="carat",
    y="price",
    alpha=0.35,
    s=20
)

plt.title("Relationship Between Diamond Carat and Price")
plt.xlabel("Carat Weight")
plt.ylabel("Price")
plt.tight_layout()
plt.show()
```

```python

carat_price_corr = diamonds["carat"].corr(diamonds["price"])

print(f"Correlation between carat and price: {carat_price_corr:.3f}")
```
